```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

a = Node('A')
b = Node('B')
c = Node('C')
d = Node('D')

a.next = b
b.next = c
c.next = d

temp = a
while temp:
    print(temp.data, end=" -> ")
    temp = temp.next
print("None")
```

```python
stack = []
top = -1

def push(data):
    global top

    stack.append(data)
    top += 1

    print(data, "inserted")

def pop():
    global top

    if top == -1:
        print("Stack is Empty")
    else:
        print("Deleted element:", stack[top])
        stack.pop()
        top -= 1


def peek():
    if top == -1:
        print("Stack is Empty")
    else:
        print("Top element:", stack[top])


push(10)
push(20)
push(30)

print("Stack:", stack)

peek()

pop()

print("Stack after pop:", stack)

peek()
```
